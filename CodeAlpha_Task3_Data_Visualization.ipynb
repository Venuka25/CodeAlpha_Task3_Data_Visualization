```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("CodeAlpha_Task1_Final_Dataset.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Price"], bins=10)

plt.xlabel("Book Price (£)")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Prices")

plt.show()
```

```python
rating_counts = df["Rating"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(rating_counts.index, rating_counts.values)

plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Number of Books by Rating")

plt.xticks([1, 2, 3, 4, 5])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Rating"], df["Price"])

plt.xlabel("Rating")
plt.ylabel("Book Price (£)")
plt.title("Book Price vs Rating")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["Price"])

plt.ylabel("Book Price (£)")
plt.title("Box Plot of Book Prices")

plt.show()
```

```python
top_10_expensive = df.sort_values("Price", ascending=False).head(10)

plt.figure(figsize=(10, 6))

plt.barh(top_10_expensive["Title"], top_10_expensive["Price"])

plt.xlabel("Book Price (£)")
plt.ylabel("Book Title")
plt.title("Top 10 Most Expensive Books")

plt.gca().invert_yaxis()

plt.show()
```

## Key Findings

- The price histogram shows the distribution of book prices across the dataset.
- The rating bar chart shows that all ratings from 1 to 5 are represented.
- The scatter plot shows the relationship between book price and rating.
- The box plot shows the spread of book prices and helps identify possible outliers.
- The top 10 chart highlights the most expensive books in the dataset.
- These visualizations make the dataset easier to understand and compare.
